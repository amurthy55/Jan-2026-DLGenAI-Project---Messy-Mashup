```python
import torch
import torch.nn as nn
import torchaudio
import math
```

```python
# Q1: Tensor flattening
# Input shape: (32, 1, 64, 64) - batch, channels, height, width
# Flatten excluding batch dimension: 1 * 64 * 64 = 4096

tensor = torch.randn(32, 1, 64, 64)
flattened = tensor.view(tensor.size(0), -1)
print(f"Q1: Input features for Linear layer: {flattened.size(1)}")  # 4096
```

```python
# Q2: Mel-Spectrogram shape
# 1-second audio at 16kHz, n_fft=400, hop_length=160, n_mels=64

sample_rate = 16000
duration = 1.0
audio = torch.randn(int(sample_rate * duration))

mel_spec = torchaudio.transforms.MelSpectrogram(
    sample_rate=sample_rate,
    n_fft=400,
    hop_length=160,
    n_mels=64
)(audio)

print(f"Q2: Mel-Spectrogram shape: {mel_spec.shape}")  # (1, 64, 101)
```

```python
# Q3: CNN output shape after Conv2d + MaxPool2d
# Input: (1, 64, 64)
# Conv2d(kernel=3, stride=1, padding=1) -> (64, 64)
# MaxPool2d(kernel=2, stride=2) -> (32, 32)

input_tensor = torch.randn(1, 64, 64)
conv = nn.Conv2d(1, 1, kernel_size=3, stride=1, padding=1)
pool = nn.MaxPool2d(kernel_size=2, stride=2)

conv_out = conv(input_tensor)
pool_out = pool(conv_out)

print(f"Q3: Spatial output shape (H, W): {pool_out.shape[-2:]}")  # (32, 32)
```

```python
# Q4: Linear layer parameters
# nn.Linear(in_features=128, out_features=64)
# Weights: 128 * 64 = 8192
# Biases: 64
# Total: 8192 + 64 = 8256

layer = nn.Linear(in_features=128, out_features=64)
total_params = sum(p.numel() for p in layer.parameters())
print(f"Q4: Total learnable parameters: {total_params}")  # 8256
```

```python
# Q5: Batches per epoch
# 1050 files, batch_size=32, drop_last=False
# 1050 / 32 = 32.8125 -> 33 batches

total_files = 1050
batch_size = 32
batches = math.ceil(total_files / batch_size)
print(f"Q5: Batches per epoch: {batches}")  # 33
```

```python
# Q6: CrossEntropyLoss calculation
# 3-class classification, logits [2.5, 1.0, 0.1], target = Class 0

logits = torch.tensor([[2.5, 1.0, 0.1]], dtype=torch.float32)
target = torch.tensor([0], dtype=torch.long)

loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(logits, target)

print(f"Q6: CrossEntropyLoss: {loss.item():.3f}")  # 0.273
```

```python
# Q7: SGD optimizer step
# Weight = 0.5, Gradient = 0.2, Learning Rate = 0.01
# New weight = Weight - LR * Gradient = 0.5 - 0.01 * 0.2 = 0.498

weight = torch.tensor([0.5], requires_grad=True)
gradient = torch.tensor([0.2])
lr = 0.01

weight.grad = gradient
optimizer = torch.optim.SGD([weight], lr=lr)
optimizer.step()

print(f"Q7: New weight: {weight.item():.3f}")  # 0.498
```

```python
# Q8: Padding for same output size
# kernel_size=5, stride=1, want same output size
# padding = (kernel_size - 1) // 2 = (5 - 1) // 2 = 2

kernel_size = 5
padding = (kernel_size - 1) // 2
print(f"Q8: Padding value: {padding}")  # 2
```
